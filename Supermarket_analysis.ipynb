```python
import pandas as pd
df=pd.read_csv('C:\\Users\\HP\\Downloads\\archive (1)\\supermarket.csv.csv')
print('csv loaded successfully')
```

```python
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine

load_dotenv()
db_password = os.getenv('DB_PASSWORD')

engine = create_engine(f'postgresql://postgres:{db_password}@localhost:5432/supermarket_db')

df.to_sql(
    'supermarket_sales',
    con=engine,
    if_exists='replace',
    index=False,
)

print('Data successfully pushed to postgresql table!')
```
